"
The hemoglobin A1c test—also known as glycated hemoglobin, glycosylated hemoglobin, HbA1c, or simply A1c—is used to measure an individual's glucose control levels. The test shows average blood sugar levels over the past 90 days, expressed as a percentage. In addition, it can be used to diagnose diabetes mellitus.[1] Hemoglobin is a protein found exclusively in red blood cells, giving blood its bright red color. The primary role of hemoglobin is to carry oxygen from the lungs to all the cells in the body. Hemoglobin becomes glycated or coated with glucose from the bloodstream. As blood glucose levels increase, more glucose attaches to the hemoglobin protein, resulting in a higher A1c value.[2] Since red blood cells have an average lifespan of about 3 months, the A1c test measures hemoglobin levels in the bloodstream over this period, making it a reliable indicator of blood sugar control.
"
(source: Basu S, et al. Hemoglobin A1C. StatPearls. 2025. Available from: https://www.ncbi.nlm.nih.gov/sites/books/NBK549816/)

In summary for TLDR: your blood has haemoglobin, and glucose. the higher your glucose concentration, the more often random collisions occur, forming a non-enzyme catalysed (as in it happens naturally by simple collision) and the more of these "complexes" meaning like combined molecules occur, called HbA1c - Hb meaning Haemoglobin, A = Adult heamoglobin, 1c = the chromatography fraction label from its lab experiment back in 1958, just the name stuck. Also, crucially, your red blood cells/hameoglobin has a lifespan of 3 MONTHS meaning once these complexes form, a test reading can tell you how much glucose has been in your blood for the last 3 months (because most of these complexes become permanently merged), if youre diabetic and glucose is high, then more complexes would have formed, and youd have an "elevated HbA1c level" - specifically, above 6.5% was the cutoff point - wait, that wasnt a TLDR at all, but just paraphrasing into my own words, sorry, but I think the passage in the extract - I don't think I couldv'e said it better myself
